## 2. Cosine Similarity Whiteboard Activity

This activity was completed by hand, here is how to do it via Python code.

### Inputs for the problem

````python
documents = [
    "Max likes Sam.",    # D1
    "Sam likes Max.",    # D2
    "Max and Sam.",      # D3
    "Winter likes fall." # D4
]

query = "Sam likes winter"
````

### Step 1: Create vocabulary of all terms

First I will concatenate all strings into one.

````python
allText = " ".join(documents + [query])
print(allText)
````

I don't want to have punctuation. There are several ways to do that. Let's see some:

#### Method 1: Iterate over the string, leave out punctuations

````python
import string
print(string.punctuation)
````

````python
allText_noPunct = "".join([ch for ch in allText if not ch in string.punctuation])
print(allText_noPunct)
````

### Method 2: Use built-in methods

Pythyn's built-in `str` class has many useful methods for translating a set of characters to another one. Let's see the method in action:

````python
allText.translate(str.maketrans("", "", string.punctuation))
````

What's going on?

````python
print(str.maketrans("", "", string.punctuation))
````

We can see that all ASCII Characters from 0-32 and from 65 to 90 are missing, because those characters will be deleted.

Here is what is happening:

```
str.maketrans(from_str, to_str, delete_chars)
#              ^          ^        ^
#           1st arg    2nd arg  3rd arg
```

* 1st argument (""): The characters you want to replace. (Empty string = replace nothing)

* 2nd argument (""): The characters to replace them with. (Empty string = no replacement mapping)

* 3rd argument (string.punctuation): The characters you want to delete entirely.

### Our vocabulary

Here is a Python one-liner that takes our variable `allText_noPunct` and find the unique words in it.

````python
words = set(allText_noPunct.lower().split())
words
````

````python
voc = sorted(words)
voc
````

### Step 2: Represent each document and the query in the vector space created by these terms

#### A. Clean the documents

````python
clean_doc = []

for doc in documents:
    doc_noPunct = "".join([ch for ch in doc if not ch in string.punctuation])
    clean_doc.append(doc_noPunct.lower().split())

print(clean_doc)
````

#### B. Create the Vectors

````python
vectors = []
for i in range(len(clean_doc)):
    vector = []
    for word in voc:
        vector.append(clean_doc[i].count(word))
    vectors.append(vector)


print(vectors)
````

##### Create the query vector

````python
query_vector = []
for word in voc:
    query_vector.append(query.lower().split().count(word))

print(query_vector)
````

### Step 3: Calculate the cosine similarity between the query and each document


#### A. Define the cosine similarity formula


````python
import numpy as np
from numpy.linalg import norm

def cosineSimilarity(vec1, vec2):
    v1 = np.array(vec1)
    v2 = np.array(vec2)
    cosine = np.dot(v1, v2) / (norm(v1) * norm(v2))
    return cosine
````

#### B. Calculate the cosine similarity

````python
similarities = []
for vector in vectors:
    similarities.append(cosineSimilarity(vector, query_vector))

print(similarities)
````

### Step 4:  Rank the results based on the cosine similarity

````python
sorted_ind = np.argsort(similarities)[::-1] # Find the indices of the sorted similarities in descending order

for i in sorted_ind:
    print("D"+str(i)+":", documents[i], ":", round(similarities[i],3))
````

#### 2. Find each show's Wiki article name

Limitation: There are some wikipedia page that is the whole-series article.

````python
articles = [
    "Teach_You_a_Lesson",
    "Four_Hands,_Two_Sonatas",
    "Bloodhounds_(TV_series)",
    "Agent_Kim_Reactivated",
    "We_Are_All_Trying_Here",
    "A_Shop_for_Killers",
    "The_Wonderfools",
    "The_Legend_of_Kitchen_Soldier",
    "A_Bona_Fide_Killer",
    "In_Your_Radiant_Season"
]
````

#### 3. Collect matching pageview rows for all 30 days of September

create `dates` to represent all days in september

````python
dates = pd.date_range(
    start="2026-09-01",
    end="2026-09-30"
).strftime("%Y-%m-%d")
````

Loop through each date

````python
import csv
daily_data= []

for date in dates:
    url = f"https://analytics.wikimedia.org/published/datasets/country_project_page/{date}.tsv"
    response = requests.get(url, headers=headers)
    df = pd.read_csv(
        io.StringIO(response.text), 
        header=None, 
        names=columns, 
        sep='\t',
        quoting=csv.QUOTE_NONE
    )
    filtered_df = df[df['page_title'].isin(articles)]
    daily_data.append(filtered_df)

print(daily_data)
````

Combine the daily tables

````python
for date,data in zip(dates, daily_data):
    data['date'] = date

september_df = pd.concat(daily_data, ignore_index=True)
september_df
````

#### 4. Save those rows as a CSV or TSV.
